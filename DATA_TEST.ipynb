```
import requests
import time

# กำหนด URL ของ API ที่ต้องการดึงข้อมูล
url = 'https://gtadmin.goldtraders.or.th/wp-admin/sse_gold.php'

def fetch_live_data():
    try:
        response = requests.get(url)
        # ตรวจสอบว่าเชื่อมต่อสำเร็จหรือไม่
        if response.status_code == 200:
            data = response.json() # แปลงข้อมูลให้อยู่ในรูปแบบ JSON
            print("ข้อมูลอัปเดตล่าสุด:", data)
        else:
            print("เกิดข้อผิดพลาดในการเชื่อมต่อ:", response.status_code)
    except Exception as e:
        print("เกิดข้อผิดพลาด:", e)

# วนลูปดึงข้อมูลแบบเรียลไทม์
while True:
    fetch_live_data()
    time.sleep(5) # หน่วงเวลา 5 วินาทีก่อนดึงข้อมูลรอบถัดไป
```
